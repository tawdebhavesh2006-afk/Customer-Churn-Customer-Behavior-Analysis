```python
#Convert Total_Spend into a NumPy array.
spending = np.array(df["Total_Spend"])
spending
```

```python
#Calculate mean spending
mean_spending = np.mean(spending)
print(mean_spending)
```

```python
#Calculate median spending
median_spending = np.median(spending)
print(median_spending)
```

```python
#Calculate standard deviation
std_spending = np.std(spending)
print(std_spending)
```

```python
#Find minimum and maximum spending
minimum = np.min(spending)
maximum = np.max(spending)

print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
#Calculate 25th and 75th percentile
p25 = np.percentile(spending, 25)
p75 = np.percentile(spending, 75)

print("25th Percentile:", p25)
print("75th Percentile:", p75)
```

```python
#Compare mean spending of Churned vs Active customers
churned_spending = np.mean(df[df["Churn_Flag"] == 1]["Total_Spend"])
active_spending = np.mean(df[df["Churn_Flag"] == 0]["Total_Spend"])

print("Churned Mean Spending:", churned_spending)
print("Active Mean Spending:", active_spending)
```
