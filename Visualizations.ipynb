```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Dell\OneDrive\Documents\DATA ANALYSIS PROJECT\Customer Churn & Customer Behavior Analysis\EXCEL\DATA TO PERFORM ON SQL AND LIBRARIES.csv")
df
```

```python
status_count = df["Customer_Status"].value_counts()

sns.barplot(
    x=status_count.index,
    y=status_count.values
)

plt.title("Customer Status Distribution")
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python
churn_segment = (
    df.groupby("Customer_Segment")["Churn_Flag"]
    .mean()
    .mul(100)
)

sns.barplot(
    x=churn_segment.index,
    y=churn_segment.values
)

plt.title("Churn Rate by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
churn_plan = (
    df.groupby("Subscription_Plan")["Churn_Flag"]
    .mean()
    .mul(100)
)

sns.barplot(
    x=churn_plan.index,
    y=churn_plan.values
)

plt.title("Churn Rate by Subscription Plan")
plt.xlabel("Subscription Plan")
plt.ylabel("Churn Rate (%)")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Customer_Status",
    y="Monthly_Logins"
)

plt.title("Monthly Logins by Customer Status")
plt.xlabel("Customer Status")
plt.ylabel("Monthly Logins")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Customer_Status",
    y="Satisfaction_Score"
)

plt.title("Satisfaction Score by Customer Status")
plt.xlabel("Customer Status")
plt.ylabel("Satisfaction Score")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Customer_Status",
    y="Total_Spend"
)

plt.title("Customer Spending by Status")
plt.xlabel("Customer Status")
plt.ylabel("Total Spend")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Customer_Status",
    y="Support_Tickets"
)

plt.title("Support Tickets by Customer Status")
plt.xlabel("Customer Status")
plt.ylabel("Support Tickets")
plt.show()
```

```python
corr_columns = [
    "Age",
    "Total_Orders",
    "Average_Order_Value",
    "Total_Spend",
    "Days_Since_Last_Purchase",
    "Monthly_Logins",
    "Support_Tickets",
    "Satisfaction_Score",
    "Discount_Usage_Rate",
    "Churn_Flag"
]

corr = df[corr_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Customer Behavior Correlation Heatmap")
plt.show()
```
